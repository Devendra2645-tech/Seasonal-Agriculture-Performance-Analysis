H₀ (Null Hypothesis):
Mean crop yield is the same across seasons.

H₁ (Alternative Hypothesis):
Mean crop yield differs across at least one season.

```python
data['Season'].value_counts()
```

```python
season_groups = [
    group["Yield_Tonnes_Ha"].dropna().values
    for name, group in data.groupby("Season")
]

season_names = list(data["Season"].unique())

print(season_names)
```

```python
anova_result = stats.f_oneway(*season_groups)

print("F-statistic:", anova_result.statistic)
print("P-value:", anova_result.pvalue)
```

```python
alpha = 0.05

if anova_result.pvalue < alpha:
    print("Reject H0: Yield differs significantly across seasons.")
else:
    print("Fail to reject H0: No statistically significant difference was detected.")
```

```python
production_groups = [
    group["Production_Tonnes"].dropna().values
    for name, group in data.groupby("Season")
]

production_anova = stats.f_oneway(*production_groups)

print("F-statistic:", production_anova.statistic)
print("P-value:", production_anova.pvalue)
```

```python
if production_anova.pvalue < 0.05:
    print("Production differs significantly across seasons.")
else:
    print("No statistically significant seasonal difference was detected in production.")
```

```python
profit_groups = [
    group["Profit_INR"].dropna().values
    for name, group in data.groupby("Season")
]

profit_anova = stats.f_oneway(*profit_groups)

print("F-statistic:", profit_anova.statistic)
print("P-value:", profit_anova.pvalue)
```

```python
if profit_anova.pvalue < 0.05:
    print("Profit differs significantly across seasons.")
else:
    print("No statistically significant seasonal difference was detected in profit.")
```

```python
statistical_results = pd.DataFrame({
    "Metric": [
        "Yield",
        "Production",
        "Profit"
    ],
    "F_Statistic": [
        anova_result.statistic,
        production_anova.statistic,
        profit_anova.statistic
    ],
    "P_Value": [
        anova_result.pvalue,
        production_anova.pvalue,
        profit_anova.pvalue
    ]
})

statistical_results
```

```python
statistical_results["Significant"] = (
    statistical_results["P_Value"] < 0.05
)

statistical_results
```
